```python
# Write a program to find the largest of 3 numbers 
def largest_among_3(a,b,c):
    if a > b and a > c:
        return a
    elif b > c:
        return b
    else:
        return c
largest_among_3(11,-22,3)
```

```python
# Write a program to calculate the factorial of a number.

def factorial(num):
    fact = 1
    for i  in range(1, num +1):
        fact = fact * i
    return fact
factorial(0)
```

```python
# check whether a number is armstrong or not 
def palindrome_number(num):
    original_num = num
    reverse = 0
    while (num != 0):
        digit = num % 10
        num = num // 10
        reverse = reverse * 10 + digit
    if original_num == reverse:
        return "Armstrong Number"
    else:
        return "Not a Armstrong Number"
print(palindrome_number(123456789987654321))
```

```python
# Write a program to find the sum of digits of a number.

def sum_of_digits(num):
    result = 0
    while(num != 0):
        result = result +  num % 10
        num = num // 10
    return result
sum_of_digits(12345678910)
```

```python
#Prime Number logic 
import math
def check_prime(num):
    if num < 2:
        return False
        
    for i in range(2, math.floor(math.sqrt(num)) + 1):
        if num % i == 0:
            return False
    return True
r1 = int(input("Enter Number: 1"))
r2 = int(input("Enter Number: 2"))
result = [i for i in range(r1, r2+1) if check_prime(i)]  
print(result)
```

```python
#  Greatest common factor
def gcd(num1, num2):
    factors1  = []
    factors2 = []
    if  num1 == 0 and num2 ==0:
        return "Undefined"
    elif num1 ==  0:
        return num2
    elif  num2 == 0:
        return num1
    else:
        for i in range(1, num1+1):
            if num1 % i == 0:
                factors1.append(i)
        for i in range(1, num2+1):
            if num2 % i == 0:
                factors2.append(i)
        print(factors1)
        print(factors2)
        return max(set(factors1).intersection(factors2))
        
gcd(6, 10)
```

```python
def gcd(a, b):
    a = abs(a)
    b = abs(b)
    while b!= 0:
        a, b = b, a % b
    return a

gcd(0,5)
```
